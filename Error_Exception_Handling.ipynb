```python
a = 10
b = 20
c = 'Harry'
```

```python
#scenario 1 - valid Logic
try:
  print(a+b)
except Exception as e:
  print(f'The error as {e}')
```

```python
#scenario 2 - invalid Logic
try:
  print(a/c)
except Exception as e:
  print(f'The error as {e}')
```

```python
print(a+b)
print(a/c)
print(a*b)
```

```python
try:
  print(a+b)
  print(a/c)
  print(a*b)
except Exception as e:
  print(e)
```

```python
try:
  print(a+b)
  print(a/c)
  print(a*b)
except Exception as e:
  print(e)
finally:
  print(a*b)
```
