```python
#Predefined Arrays

#zeros
z1 = np.zeros((2,4))
print(z1)
print(type(z1))
```

```python
#ones
o1 = np.ones((2,4))
print(o1)
print(type(o1)) 
```

```python
#full 
f1 = np.full((2,4),15)
print(f1)
print(type(f1))
```

```python
#Ranges
n1 = np.arange(5,15)
print(n1)
```

```python
n2 = np.arange(5,15,2)
print(n2)
```

```python
#linspace
nln1 = np.linspace(10,20,4)
print(nln1)
```

```python
nln2 = np.linspace(1,10,10,dtype=int)
print(nln2)
```

```python
#random arrays
nrn1 = np.random.rand(2,3) #works as a dimensional matrix
print(nrn1)
```

```python
nrn2 = np.random.randint(2,30,(2,3)) #(2,30 works as a range) / (2,3 works as dimensional matrix)
print(nrn2)
```

```python
arr10 = np.array([4,8,9,15]) 
arr11 = np.array([2,6,13,17])
```

```python
print(arr10+arr11)
print(arr10-arr11)
print(arr10*arr11)
print(arr10/arr11)
```

```python
arr12 = np.array([1,4,9,16])
print(np.sqrt(arr12))
print(np.log(arr12))
print(np.exp(arr12))
```

```python
#reshape - wrong
arr13 = np.arange(1,7).reshape(3,3)
arr13
```

```python
#reshape - right
arr13 = np.arange(1,7).reshape(2,3)
arr13
```

```python
#statistical functions
print(np.sum(arr13))
print(np.mean(arr13))
print(np.std(arr13))
print(np.min(arr13),np.max(arr13))
```

```python
#dimensionality 
ar2d = np.array(
[
    [1,2,3],
    [4,5,6]
]
)
print(ar2d)
print(ar2d.ndim)
print(ar2d.shape)
```

```python
ar3d = np.array(
    [
        [
            [1,2,3],
            [4,5,6]
        ]
    ]
)
print(ar3d)
print(ar3d.ndim)
print(ar3d.shape)
```

```python
ar4d = np.ones((2,3,4,5),dtype=int)
print(ar4d)
print(ar4d.ndim)
print(ar4d.shape)
```
