```python
def fname(first_name,last_name):
  full_name = first_name+' '+last_name
  return (f'''the first name is {first_name} and the last name is {last_name} and the full name is {full_name}''')
```

```python
fname('harry','potter')
```

```python
#args and kwargs
#args - positional arguments
#kwargs - keyword arguments
```

```python
def multi_arg(*args):
  return sum(args)
```

```python
multi_arg(5,8,9,10,56)
```

```python
#kwargs
def print_details(**kwargs):
  for key,value in kwargs.items():
    print(f'{key}:{value}')
```

```python
print_details(name='Ankit',hometown='Kolkata',std = 12)
```
