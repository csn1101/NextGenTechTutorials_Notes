#### **Python List & Dictionary Comprehensions**

```python
#List comprehensions
lst1 = []
for i in range(15):
  if i%2==0:
    lst1.append(i)
```

```python
print(lst1)
```

```python
lst2 = []
for i in range(15):
  if i%2==1:
    lst2.append(i)
```

```python
print(lst2)
```

```python
lst3 = [i for i in range(15) if i%2==0]
print(lst3)
```

```python
lst4 = [i*3 for i in range(15) if i%2==0]
print(lst4)
```

```python
dict1 = {}
for i in range(10):
  if i%2==0:
    dict1[i] = i*2
```

```python
print(dict1)
```

```python
dict2 = {i: i**3 for i in range(10) if i%2 == 0}
print(dict2)
```
