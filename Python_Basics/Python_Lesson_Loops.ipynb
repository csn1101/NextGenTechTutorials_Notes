### **Python Loops**

1. For Loop
2. While Loop

#### **For Loop**

```python
lst = ['apple','banana','orange']
print(lst)
```

```python
for items in lst:
  print(items)
```

```python
for item in lst:
  print(len(item))
```

```python
for item in lst:
  print(item.upper())
```

```python
for i in range(5):
  print(i)
```

```python
for i in range(1,10):
  print(i)
```

```python
for i in range(1,10,2):
  print(i)
```

```python
students = {'name':'Alok','class':12,'marks':85}
for i in students:
  print(i)
```

```python
for i in students.items():
  print(i)
```

```python
for i,j in students.items():
  print(f'the key is {i}, and the value is {j}')
```

#### **While Loop**

```python
a = 4
while a>=3:
  print('This loop is running')
  a = a-1
```
