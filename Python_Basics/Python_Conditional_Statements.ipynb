#### Conditional Statements In Python

```python
#if - else blocks
a = 16
if a>15:
  print('a is good')
else:
  print('a is bad')
```

```python
marks = 30
if marks>=34:
  print('Pass')
else:
  print('Fail')
```

```python
#else if/ elif statements
marks = 72
if marks>=70:
  print('Grade A')
elif marks>=60:
  print('Grade B')
elif marks>=50:
  print('Grade C')
elif marks>=40:
  print('Grade D')
else:
  print('Fail')
```

```python
#Nested Conditional Statements
a = -1
if a>0:
  print('The number is positive')
  if a%2==0:
    print('The number is even')
  else:
    print('The number is odd')
else:
  print('The number is negative')
```

```python
lst1 = [15,25,45,30]
for i in lst1:
  if i>40:
    print(i)
  else:
    print('not sure')
```
