```python
#Lists
# we define lists with []
#integer list
a = [10,25,36,85]
print(a)
print(type(a))
```

```python
#Lists can contain multiple data types
#string list
b = ['apple','banana','blueberry']
#floating point numbers
c = [25.58,10.5,45.96]
print(type(b))
print(type(c))
```

```python
#A list contain mixed data type elements
d = ['apple',25,10.50]
print(d)
print(type(d))
```

```python
#accessing and slicing list
#accessing the list
print(d[2])
```

```python
#sliced the list
lst = ['apple','banana','blueberry',15,44,89,52.53,10.45]
print(lst[1:5])
```

```python
#negative indexing also works with list
print(lst[-3])
```

```python
#step function in lists
print(lst[0:6:2])
print(lst[-5:-1:2])
```

```python
#Lists are mutable
#List elements can be changed
print(lst)
```

```python
lst[0] = "orange"
print(lst)
```

```python
#append method for adding elements
lst.append("mango")
print(lst)
```

```python
#insert method
lst.insert(3,"grapefruit")
```

```python
print(lst)
```

```python
#remove the elements from a list
#remove method - providing element's name
lst.remove('orange')
print(lst)
```

```python
#pop method
lst.pop()
print(lst)
```

```python
var1 = lst.pop()
print(var1)
```

```python
#del method
del lst[1]
print(lst)
```

```python
#clear metod - this will clear all the items from the list but the list will remain
lst.clear()
print(lst)
```
