```python
#interger type numbers
a = 10
b = 20
print(a)
print(type(a))
```

```python
#floating point number
b = 10.56
print(b)
print(type(b))
```

```python
#complex type of numbers
# real part of the number
# imaginary part of the number
c = 3+4j
print(c)
print(type(c))
```

```python
print(c.real)
print(c.imag)
```

```python
#imaginary part of the number provides you a squared value in negative 1
d = 2*2
print(d)
```
