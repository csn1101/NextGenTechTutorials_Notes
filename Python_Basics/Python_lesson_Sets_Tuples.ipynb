### Python Sets & Tuples

##### **Tuples**

```python
#Tuples
empty_tpl = ()
print(empty_tpl)
print(type(empty_tpl))
```

```python
tpl_1 = (12,"horse",15.5,True)
print(tpl_1)
print(type(tpl_1))
```

```python
#single item tuple constraint
tpl_2 = ("elephant")
print(tpl_2)
print(type(tpl_2))
```

```python
#if we want to create a tuple with a single element , we must provide a ","
tpl_3 = ("elephant",)
print(tpl_3)
print(type(tpl_3))
```

```python
#nested tuples
tpl_4 = (1,2,(3,4,("park",56.5)))
print(tpl_4)
print(type(tpl_4))
```

```python
#accessing and slicing tuples
print(tpl_4[0])
print(tpl_4[1])
print(tpl_4[2][2][0])
```

```python
print(tpl_4[0:2])
```

```python
print(tpl_4[-1])
```

```python
num_tpl = (1,2,9,11,4,5,6,3)
print(num_tpl.count(3))
```

```python
print(num_tpl.index(3))
```

```python
print(len(num_tpl))
```

```python
fruits = ('apple','banana','cherry')
```

```python
#concatenate
fruits_1 = fruits + ("orange","kiwi")
print(fruits_1)
```

```python
fruit_number = fruits_1 + num_tpl
print(fruit_number)
```

##### **Set**

```python
empty_set = set()
print(empty_set)
print(type(empty_set))
```

```python
set_1 = set({12,"apple",'banana'})
print(set_1)
print(type(set_1))
```

```python
set_2 = {14,"cherry",'banana'}
print(set_2)
print(type(set_2))
```

```python
#wrongly created set attempt
set_3 = {}
print(type(set_3))
```

```python
set_10 = {1,2,3,4}
set_11 = {4,9,3,10}
```

```python
print(set_10.union(set_11))
```

```python
print(set_10.intersection(set_11))
```

```python
print(set_10.difference(set_11))
print(set_11.difference(set_10))
```

```python
set_10.add(15)
print(set_10)
```

```python
set_10.remove(3)
print(set_10)
```

```python
set_10.pop()
print(set_10)
```
