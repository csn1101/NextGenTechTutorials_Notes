#### Python Dictionaries

Key (The word you meant to ask the meaning) - value(meaning itself)
- Dictionaries as key value pairs
-  Example - name - Suraj Kumar
- Mutable - The value of a key can be changed
- Keys should be unique
- This data type is unordered
- {"name":"Chandan" , "class":"12"}

```python
empty_dict = {}
print(empty_dict)
print(type(empty_dict))
```

```python
# If you provide any value other than key value pair within the dictionary brackets
# It will not remain a dictionary
dict1 = {"name","class"}
print(dict1)
print(type(dict1))
```

```python
dict2 = {"name":"Chandan","class":12}
print(dict2)
print(type(dict2))
```

```python
#accessing the elements of dictionary is equally important
#method 1
print(dict2['name'])
#method 2
print(dict2.get('class'))
```

```python
print(dict2.get('hometown','Not Available'))
```

```python
#updating and adding elemnts into a dictionary
dict2['class'] = 10
print(dict2)
```

```python
dict2['hometown'] = "Kolkata"
print(dict2)
```

```python
# Removing an elemnt from dictionary
# Pop removes the element by it's key
dict2.pop('hometown')
print(dict2)
```

```python
#popitem removes the last element in your dictionary
dict2.popitem()
print(dict2)
```

```python
#Dictionary Methods
temp = {'a':1,'b':2,'c':3}
print(f'shows all keys in dictionry {temp.keys()}')
print(f'shows all values in dictionry{temp.values()}')
print(f'shows all items in dictionry{temp.items()}')
```

```python
temp2 = temp.copy()
print(temp2)
```

```python
temp2.clear()
print(temp2)
```

```python
students = {'name':'harry','hometown':'england','age':25}
print(students)
```

```python
#updating with a dictionary
update_dict = {'hometown':'ireland','age':22}
```

```python
students.update(update_dict)
print(students)
```

```python
#nested dictionaries
college_database = {
    'college1':{'name':'havard','student_count':12500},
    'college2':{'name':'cambridge','student_count':32500},
    'college3':{'name':'IIT','student_count':12550}
}
print(college_database['college3'])
```
