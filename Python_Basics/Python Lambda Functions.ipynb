#### **Python Lambda Functions**

```python
def avg(a,b):
  c = (a+b)/2
  return c
```

```python
print(avg(6,8))
```

```python
avg1 = lambda x,y:(x+y)/2
```

```python
print(avg1(6,9))
```

```python
multi1 = lambda x: x*4
```

```python
print(multi1(4))
```

```python
avg3 = lambda x,y,z:(x+y+z)/3
```

```python
print(avg3(15,45,96))
```

```python
square = lambda x: x*x
print(square(4))
```

```python
cube = lambda x: x*x*x
print(cube(3))
```

```python
def func1(fx,x):
  return(fx(x)/2)
```

```python
print(func1(cube,5))
```
