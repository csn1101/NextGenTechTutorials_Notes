#### String Formatting

- legacy method - %d,%s,%f
  - d - integer decimal - 25
  - s - string - "hello"
  - f - floating number - 25.50
- str.format
- f string method

```python
a = 25
b = "hello"
c = 25.50
#this is not dynamic as we are hard coding the value , string formatting helps in fixing the same.
print("the value of a is 25")
```

```python
print("the string value in our data is %s , integer value as %d , float as %f"
      % (b,a,c))
```

```python
#str.format methods
#method 1
print("the value of a is {}".format(a))
name = 'harry'
surname = 'potter'
#method 2
print("the name of the person is {f} {s}".format(f=name,s=surname))
```

```python
#f string method
x = "this is India"
y = 1.39
print(f"I am in a great country ,{x} , the population is {y} billion people")
```

##### String Functions

```python
txt = "Python is a Great Programming language"
name = "harry potter"
```

```python
print(f"the length of txt is {len(txt)}")
print(f"lowercase conversion of txt is - {txt.lower()}")
print(f"uppercase conversion of txt is - {txt.upper()}")
print(f"the capitalized version of name is - {name.capitalize()}")
print(f"the proper capitalized version of name is - {name.title()}")
print(f"the replaced version of the txt is - {txt.replace('Python','Java')}")
print(f"the split version of the txt - {txt.split()}")
print(txt.startswith('Python'))
print(txt.startswith('C++'))
print(f"the count of the letter a in the txt is {txt.count('a')}")
```

```python
#find in strings
print(txt.find('G'))
```

```python
#find in strings
print(txt.find('Awsome'))
```
