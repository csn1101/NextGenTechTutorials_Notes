#### **Abstraction**

```python
from abc import ABC, abstractmethod
```

```python
class Sound(ABC):
  @abstractmethod
  def sound(self):
    print('sound')
```

```python
class Animal:
  def meow(self):
    print("Cat Sound")
```

```python
cat = Animal()
```

```python
cat.meow()
```

```python
guitar = Sound()
```

```python
class Music(Sound):
  def sound(self):
    pass
```

```python
guitar = Music()
```

```python
guitar.sound()
```

#### **Encapsulation**

```python
class Bankaccount:
  def __init__(self,balance):
    self.__balance = balance #private variable
  def deposit(self,amount):
    self.__balance += amount
  def withdrawal(self,amount):
    self.__balance -= amount
  def get_balance(self):
    return self.__balance
```

```python
account = Bankaccount(1000)
```

```python
account.get_balance()
```

```python
account.deposit(1500)
```

```python
account.get_balance()
```

```python
account.withdrawal(250)
```

```python
account.get_balance()
```

```python
an = Animal1()
an2 = Animal2()
```

#### **Inheritence**

```python
class Grandfather:
  def House(self):
    print('I own The House')
```

```python
Grandpa = Grandfather()
Grandpa.House()
```

```python
class Father(Grandfather):
  def Car(self):
    print('I own The Car')
```

```python
Dad = Father()
Dad.Car()
```

```python
Dad.House()
```

```python
# The inheritence does not work upwards unless assigned
Grandpa.Car()
```

```python
class Child(Father):
  def iPhone(self):
    print('I own an Iphone')
```

```python
son =  Child()
```

```python
son.House()
```

```python
son.Car()
```

```python
son.iPhone()
```

```python
class Dad1:
  def bike(self):
    print('bike')
class Mom1:
  def jewellery(self):
    print('necklace')
```

```python
# A child can inherit properties from multiple parents
class son1(Dad1,Mom1):
  def cycle(self):
    print('cycle')
```

```python
rohit = son1()
```

```python
rohit.bike()
```

```python
rohit.jewellery()
```

```python
rohit.cycle()
```
