#### **Modes in File Handling**

```python
# r mode - Read Mode
import os
file_path = r'..\data\sample.txt'
```

```python
if os.path.exists(file_path):
    with open(file_path,'r') as f:
        print(f.read())
else:
    print('file not found')
```

```python
target_line = 3
with open(file_path,'r') as f:
    for current_line,content in enumerate(f,start=1):
        if current_line == target_line:
            print(content)
            break
```

```python
#write mode
#with write mode if the file doesn't exist then it creates the file
#it overwrites the existing content
```

```python
file_path1 = r'..\data\sample1.txt'
with open(file_path1,'w') as f:
    f.write("This is line 1 ... \n")
    f.write("This is line 2 ... \n")
```

```python
with open(file_path,'w') as f:
    f.write("This is over writing the existing file ... \n")
```

```python
#append mode
with open(file_path,'a') as f:
    f.write("this line is getting appended")
```

```python
#r+ method
with open(file_path,'r+') as f:
    data = f.read()
    f.seek(0)
    f.write("this is the updated starting point")
```

```python
#a+ - appending and reading
with open(file_path,'a+') as f:
    f.write("\n.appended line \n")
    f.seek(0)
    print(f.read())
```

```python
#Handling Binary Files - rb
binary_file_path = r'..\data\messi.jpg'
```

```python
with open(binary_file_path,'rb') as f:
    binary_data = f.read()
```

```python
from PIL import Image
from io import BytesIO
from IPython.display import display
```

```python
#convert binary data into a file like object
image_stream = BytesIO(binary_data)
#opening the image from stream
img = Image.open(image_stream)
#display the image
display(img)
```
